2.  What is the difference between Hard Link & Soft Link?
  - Hard Link:
        -A hard link is a direct reference to a file's data on hard drive. Think of it as an alternative name for the same file. When you create a hard link, you essentially creating a new directory entry that points to the same underlying inode.
        -The inode is a data structure that contains all the information about a file, such as it's location on disk,permissions and timestamps.
          -Identical Inode: Hard links shares the exact same inodes number as the original files.
          -Data Integrity: If you delete the original file, the hard link still works because it points directly to the data. The files's data is only deleted when all hard links to it are removed.
          -Limitations: Hard links cannot reference directories and cannot span across different file systems or disk partitions.
          -Command: ln <original_file> <link_name>

   - Soft Link: A soft link is a shortcut or pointer to another file or directory by it's name. Unlike a Hard Link, a soft link is a new, separate filewith it's own unique inode number.
   -This new file simply contains the path to the original file.
     -Different Inodes: Soft links have a different inode number from the original file.
     -Broken Links: If the original file is moved, renamed, or deleted, the soft link will "break" and become a dangling link that points to a non-existent location.
     -Location: Soft links can reference directories and can span across different file systems and disk partitions.
     -Command: ln -s <original_file_or_directory> <link_name>

3.  What is a Kernel in Linux?
    - A Kernel is the core component of a Linux operating system. It's the central programe that manages all the fundamental operations of the computer's hardware and software.
    - It acts the "brain" of the OS, acting as an intermediary between the hardware components (CPU, memory, storage) and software applications to use.
    - It's key functions include Process management, Memory management, Device Management, System calls and file system management.

  8. You suspect that a particular process is consuming excessive CPU resources on your Linux server. How would you identify and terminate this process?
    
     - To identify and terminate a process consuming excessive CPU resources on a Linux server, we should do couple of things given below:
     -Identify the Process: by using top command

            top

    - This shows a real-time view of CPU usage.
    - The most CPU-intensive processes appear at the top.

    - or by using: ps

           ps aux --sort=-%cpu | head -10
    - This shows the top 10 CPU-consuming processes.
  
  -Terminate the Process:
      - by using a Graceful Termination

           kill <PID>
     
      - or by using Force Termination

           kill -9 <PID>
     
     - Verify the Process Is Gone

           ps -p <PID>
